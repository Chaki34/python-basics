```python
def func(input_list):

    # Separate lists for each data type
    str_list = []
    number_list = []      # Stores both int and float
    list_list = []
    dict_list = []
    tuple_list = []
    set_list = []

    #  isinstance this help to comare the instace type of the datatypes with expectied datatype

    # Traverse the input list
    for i in input_list:

        if isinstance(i, str):
            str_list.append(i)

        elif isinstance(i, (int, float)):
            number_list.append(i)

        elif isinstance(i, list):
            list_list.append(i)

        elif isinstance(i, dict):
            dict_list.append(i)

        elif isinstance(i, tuple):
            tuple_list.append(i)

        elif isinstance(i, set):
            set_list.append(i)

        else:
            return -1

    return {
        "String": str_list,
        "Numbers": number_list,
        "List": list_list,
        "Dictionary": dict_list,
        "Tuple": tuple_list,
        "Set": set_list
    }


# Input
l = [
    "debmalya",
    12,
    13.4332,
    [1, 4, 3, 2],
    {10: "devi", 20: "devid"},
    (10, 20, 30),
    {10, 20, 30, 32}
]

# Function Call
result = func(l)

# Print Result
for key, value in result.items():
    print(f"{key}: {value}")
```
