```

# if we need to access a instance method into static method and also vise varsa

class Employee2:

    def employeeDetails(self):
        print("Employee Name: Debmalya")
        print("Company: GOOGLE")

    @staticmethod
    def callInstanceMethod(employee):
        print("Inside Static Method")

        # Static method does not have self.
        # So we explicitly receive the employee object.
        employee.employeeDetails()


e1 = Employee2()

# Calling static method
Employee2.callInstanceMethod(e1)

```
