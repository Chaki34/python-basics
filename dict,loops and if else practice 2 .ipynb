```

li = ['dev',10,23,'devid',34]

# requerment is  store the int ans str into separate new lists

new_int = []
new_str = []


for  i  in  li :
  if(type(i) == int):
    new_int.append(i)
  elif (type(i) == str):
    new_str.append(i)

print("int list :",new_int)
print("str list :",new_str)
```
