```
# reduce funtion

# task

l = [10,20,30,40]
# sum of totla item of l
sum = 0
for  i in l :
  sum  = sum + i
print(sum)


# using reduce function

from functools import  reduce

reduce (lambda x , y : x + y , l) # also do same work

```
