```
# project - Using mUlti threading extract text and sotre into there folders

import urllib.request as urllib
import threading

def file_Download(url, sotredLOcation):
  urllib.urlretrieve(url, sotredLOcation)


file_Download('https://www.gutenberg.org/files/1342/1342-0.txt','sample.txt')



```

```
# url links

links = [

        "https://www.gutenberg.org/files/1342/1342-0.txt",



        "https://www.gutenberg.org/files/11/11-0.txt",



        "https://www.gutenberg.org/files/84/84-0.txt",


        "https://www.gutenberg.org/files/1661/1661-0.txt",


        "https://www.gutenberg.org/files/25344/25344-0.txt",

]
```

```
# sotring locations

file_store=['test1.txt','text2.txt','text3.txt','test4.txt','test5.txt','test5.txt']


threads = [
    threading.Thread(
        target=file_Download,
        args=(links[i], file_store[i])
    )
    for i in range(len(links))
]
```

```

# Start all threads
for t in threads:
    t.start()

# Wait for all threads to finish before continuing
for t in threads:
    t.join()
```
