```
# logging  tutorial

"""
### Logging Module in Python

The `logging` module is used to record messages about the execution
of a program, such as debugging information, warnings, and errors.

It is better than print() because it provides log levels, timestamps,
and control over which messages are displayed.

### Logging Hierarchy

NOTSET (0)
   ↓
DEBUG (10)
   ↓
INFO (20)
   ↓
WARNING (30)
   ↓
ERROR (40)
   ↓
CRITICAL (50)

### Log Levels

NOTSET    → No specific level; inherits from parent logger.
DEBUG     → Detailed information useful for debugging.
INFO      → Confirms normal program execution.
WARNING   → Something unexpected happened, but program can continue.
ERROR     → A serious problem occurred; an operation failed.
CRITICAL  → Very serious error that may stop the program.

### Basic Syntax

logging.basicConfig(level=LEVEL)
logging.LEVEL("message")

Example:
logging.info("Program started")

### Key Point

The logging level acts as a filter.
If level = INFO, then INFO, WARNING, ERROR and CRITICAL
messages are displayed, while DEBUG messages are ignored.
"""


import logging
import os

logging.basicConfig(filename="testlogfile.log",level=logging.DEBUG,force=True) # for here need this  force=True


print(os.path.abspath("testlogfile.log"))

logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")




```

```
# task  implent logging for this system

import random
import string

import logging

class  student :



    logging.basicConfig(
      filename="student.log",
      level=logging.DEBUG,
      format="%(asctime)s - %(levelname)s - %(message)s",
      force=True
      )

    def __init__(self,name,email_id,student_dept):
        logging.info("Program started")
        self.id = (
    random.choice(string.ascii_uppercase)
    + str(random.randint(1, 100))
    + random.choice(string.ascii_lowercase)
     )
        logging.critical("Id gerated : " +self.id )
        self.name = name
        self.email_id = email_id
        self.student_dept = student_dept

    def return_student_details(self):
        logging.info("return student details for id  " + self.id)
        return self.id,self.name,self.email_id,self.student_dept



    def cal_marks(self, marks_list):
     logging.info("Get The marks dict")
     logging.info("calculation........")
     total = sum(marks_list.values())
     logging.info("Total marks : " + str(total))
     logging.info("Program Ended & return to id  : "+self.id)
     return total



print("----- WELCOME TO OUR SYSTEM -----")

name = input("Enter your name: ")

# Email validation
while True:
    email_id = input("Enter your email id: ")

    if "@" in email_id and email_id.endswith(".com"):
        break
    else:
       logging.error("Invalid email! Email must contain '@' and end with '.com'.")
       print("Invalid email!")


# Department validation
valid_depts = ["CSE", "EEE", "ECE", "BBA", "BCA"]

while True:
    student_dept = input("Enter your department: ").upper()

    if student_dept in valid_depts:
        break
    else:
        logging.error("Invalid department!")
        print("Valid departments:", valid_depts)


student1 = student(name, email_id, student_dept)

print("Details:\n", student1.return_student_details())


print("----- WELCOME TO OUR MARKS CALCULATION SYSTEM -----")

dict_makrs = {}

while True:
    subject = input("Enter subject (0 to stop): ")

    if subject == "0":
        break

    marks = int(input("Enter marks: "))

    dict_makrs[subject] = marks


print("Marks:", dict_makrs)

total_marks = student1.cal_marks(dict_makrs)

print("Total Marks:", total_marks)



# print the log file msgs
with open("student.log","r") as f :
    print()
    print(f.read())
    print()







```
