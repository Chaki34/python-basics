```
# Property Decorators - Getters, Setters & Deleters
#
# Property decorator is used to control access to private attributes.
#
# property  -> Getter
# @name.setter -> Setter
# @name.deleter -> Deleter
#
# With property decorators, we can access methods like attributes.
# We don't need to call getter/setter/deleter methods using ().


class Student:

    def __init__(self, name, phone):
        self.__name = name       # private
        self.__phone = phone     # private

    # Getter
    # property allows us to access the method like an attribute.
    # Example: a.name
    @property
    def name(self):
        return self.__name

    # Setter
    # @name.setter is used to change/update the private __name.
    # Example: a.name = "Debashish"
    @name.setter
    def name(self, name):
        self.__name = name

    # Getter for phone
    @property
    def phone(self):
        return self.__phone

    # Setter for phone
    @phone.setter
    def phone(self, phone):
        self.__phone = phone

    # Deleter
    # @name.deleter is used to delete the private __name.
    # Example: del a.name
    @name.deleter
    def name(self):
        del self.__name
        print("successfully deleted")


a = Student("Deb", 333)

# Getter
print("Original Name:", a.name)

# Setter
a.name = "Debashish"

# Getter after setting new value
print("Set data:", a.name)

# Deleter
del a.name

# del a.name   # now not possible

a.name = "amit"
print("Set data:", a.name)



# After deleting a.name, accessing a.name will raise AttributeError.
# print(a.name)
```
