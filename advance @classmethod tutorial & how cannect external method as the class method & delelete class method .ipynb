```
# ============================================================
# Using an external method as a class method
# ============================================================

class Teacher:

    # Normal constructor
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
        self.score = 23

    # --------------------------------------------------------
    # Alternative constructor using @classmethod
    # --------------------------------------------------------
    @classmethod
    def from_string(cls, input_str):
        # Example input: "Deb-12233"
        name, phone = input_str.split("-")

        # cls(name, phone) creates a Teacher object
        # and internally calls __init__()
        return cls(name, phone)

    # Normal instance method
    def print_details(self):
        print(self.name, self.phone, self.score)


# Creating an object using the class method
obj = Teacher.from_string("Deb-12233")

# Calling the instance method
obj.print_details()

# Accessing an instance attribute
print(obj.phone)

 #obj.details() not rechable and not show alos hint


# ============================================================
# Adding an external function to the class as a class method
# ============================================================

# This function is defined OUTSIDE the Teacher class.
#
# The first parameter 'cls' will receive the class itself
# when this function is converted into a class method.
def details(cls, course_name):
    print("course name is  :",course_name)
    print("class name is    :",cls.__name__)


# Attach the external function to the Teacher class.
#
# classmethod(details) converts the external function into
# a class method.
Teacher.details = classmethod(details)


# Now 'details' behaves like a class method.
#
# When we call obj.details(123), Python automatically passes
# the Teacher class as the first argument (cls).
obj.details(123)
```

```
# if we want to delete any class method

del Teacher.print_details
# print(obj.print_details()) not rechable and not show alos hint

del Teacher.from_string

```
