```
# Lambda Function

# if we have a task to take n any number and p any number and return n two the pow p (n**p)

def countpow (a,b):
  return  a ** b

countpow(10,2)
```

```
# using lambda funtion we make this funtion in strightline in easier way

a = lambda a,b: a ** b # a is named entity  using this we call the lambda fun

a(10,2)
```

```
# lambda fun use for simple use  , for complex we go with treditional approche

# task find farenhite to celcious

c_to_f = lambda c : (9/5)*c + 32

c_to_f(13)
```

```
# find max b/w two numbers usong lambda fun

find_max = lambda a , b : a if a>b  else b

find_max(10,8)
```

```
# return len of str

len_str = lambda string : len(string)

len_str("durgesh")
```
