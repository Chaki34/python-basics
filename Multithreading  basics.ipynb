```
# Multithreading
#
# A process is a program that is being executed.
# A process can contain one or more threads.
#
# The Operating System (OS) manages processes and threads
# and decides which CPU core will execute them.
#
# On a multi-core CPU:
#   Dual-core  -> 2 cores
#   Quad-core  -> 4 cores
#   Octa-core  -> 8 cores
#
# Multiple processes can execute on different CPU cores.
#
# A thread is a smaller unit of execution inside a process.
# Multiple threads can be created inside the same process.
#
# IMPORTANT:
# Python's threading is especially useful for I/O-bound tasks
# such as file operations, network requests, waiting, etc.
#
# For CPU-bound tasks, Python's normal threads are limited by
# the GIL in CPython, so multiple threads do not normally give
# true parallel CPU execution for Python code.
#
# The OS, not us directly, decides which CPU core executes
# each thread.
#
# ------------------------------------------------------------
# TASK:
# Print multiplication tables from 1 to 10 using threads.
# ------------------------------------------------------------

import threading
import time


def tablePrint(number):

    # Print the multiplication table from 1 to 10
    for i in range(1, 11):

        print("Table number : %d for Place %d \n" % (number,i))


        print("%d x %d = %d" % (number, i, number * i))
        print()


        # Pause for 1 second
        # This simulates a task that spends time waiting.
        #time.sleep(1)




# Create one thread for each multiplication table.
#
# num will be:
# 1, 2, 3, ... 10
#
# args=(num,) passes the current number to tablePrint().
threads = [
    threading.Thread(target=tablePrint, args=(num,))
    for num in range(1, 6)
]


# Start all threads
for t in threads:
    t.start()


# Wait for all threads to finish
#
# Without join(), the main thread does not explicitly
# wait for these threads to complete.



print("All tables completed.")
```
