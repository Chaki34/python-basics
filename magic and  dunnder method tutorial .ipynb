```
# Special Magic Methods / Special Dunder Methods
#
# Special methods are predefined methods in Python that have a
# double underscore (__) at the beginning and at the end of their name.
#
# They are also called "Dunder Methods" because Dunder means
# "Double Underscore".
#
# These methods are not normally called directly by the programmer.
# Python automatically calls them when we perform certain operations
# on an object.
#
# They allow us to define or customize how an object behaves
# with built-in Python operations.
#
# Examples of operations where Python uses special methods:
#
# Object creation
# Object representation / printing
# Addition
# Comparison
# Length checking
# Object deletion
#
# Some commonly used special methods are:
#
# __init__()  -> Initializes an object when it is created.
# __str__()   -> Defines the user-friendly string representation of an object.
# __repr__()  -> Defines the official/developer representation of an object.
# __add__()   -> Defines the behavior of the + operator.
# __eq__()    -> Defines the behavior of ==.
# __lt__()    -> Defines the behavior of <.
# __len__()   -> Defines the behavior of len().
# __del__()   -> Called when an object is about to be destroyed.
#
# The main purpose of special/dunder methods is to allow our
# custom objects to work naturally with Python's built-in operations.
#
# In simple words:
#
# Special Dunder Methods = Special methods that Python automatically
# calls to perform or customize built-in operations on objects.


print(dir(str))

print(dir (int))

a  = 100

print("result is  :",a.__add__(5))
```
