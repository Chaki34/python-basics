```
from re import X
# Filter funtion

# task find only wich are div by 5 from a list

l = [10,23,25,34,232,65,45,76,67,8,9,0,2]
new_list = []
for i in l :
  if i % 5 == 0:
    new_list.append(i)

print(new_list)


# using filter function

new_list = lambda x : x % 5 == 0

list(filter(new_list,l)) # do same thing
```

```
#  task find negative number from a list

list1 = [1,-2,4,-5]

list(filter(lambda x : x < 0 , list1))
```

```
# task find count occurarence hello  from a long
#string

string = "hello world hello python hello programming hello AI HELLO machine learning hello data science hello coding hello developer hello software engineering hello algorithms hello databases hello networking hello cloud computing hello cybersecurity hello web development hello backend hello frontend hello full stack hello Java hello Spring Boot hello React hello Docker hello Kubernetes hello Git hello GitHub hello OpenAI hello ChatGPT hello everyone hello!"

words = string.split()

count = 0

for word in words:
    if word == "hello" or word == "HELLO":
        count += 1

print(count)

# using  filter function

hello_str = lambda x : x == "hello" or x == "HELLO"



print(filter(hello_str,string)) # filter object

result = list(filter(hello_str, string.split()))


print(result)
print(len(result))

```
