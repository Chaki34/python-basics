```
# constractor  - use when we need to set value  in the class fileds
# in time of object cration , we use constractor that injects the values of the fileds  when crate the obj

# basically in provious without constructor  we no need to pass the values of filerd thorugh the parameter we set the values in the class filed using objet that we crate  but using costructor  we can pass values in time of object crateion so obj crate for separate times for each requset



# task  make a class that hold a student entry details

import uuid

class StudentEntry:

    # constructor
    def __init__(self, name, address, phoneno, sub):
        self.id = uuid.uuid4()
        self.name = name
        self.address = address
        self.phoneno = phoneno
        self.sub = sub


    def welcomemsg(self):
     print("{")
     print(f'  "id": "{self.id}",')
     print(f'  "name": "{self.name}",')
     print(f'  "address": "{self.address}",')
     print(f'  "phoneno": "{self.phoneno}",')
     print(f'  "subjects": {self.sub}')
     print("}")



obj = StudentEntry("Debmalya","kol 700129","7001293399",["beg","eng","math","bio","phy","chmy"]) # obj of deb


obj.welcomemsg()

obj2 = StudentEntry("Sumona","kol 700129","7001293399",["beg","eng","Geo","bio","phy","history"]) # obj of sumona

obj2.welcomemsg()
```

```
# access filed instances

print(obj.sub)

print(obj2.sub)

print(obj.name)
print(obj.address)
```
