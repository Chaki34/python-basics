```
# Exception Hnadeling

# task

num1=input("Enter your number  :")
num2=input("enter your second number :")

try:
  print("Result is  :",int(num1)/int(num2))

except Exception as e:
  print("EXCPTION NOT DEVIDE BY 0 : >>> ",e)

else:
  print("Wow MY try block executed")

```

```
# ============================================================
# 1. TRY BLOCK
# ============================================================
# The try block contains the code that might cause an exception.
# Python first executes the statements inside the try block.

try:
    # Open "test.txt" in write ("w") mode.
    # If the file does not exist, Python creates it.
    # If it already exists, its previous content is overwritten.
    file = open("test.txt", "w")

    # Take input from the user.
    name = input("Enter your name: ")

    # Write the user's name into the file.
    # "\n" moves the cursor to the next line.
    file.write(name + "\n")


# ============================================================
# 2. EXCEPT BLOCK
# ============================================================
# The except block executes ONLY when an exception/error
# occurs inside the try block.
#
# "Exception as e" stores the actual error object in variable e.
# We can print e to see what went wrong.

except Exception as e:
    print("EXCEPTION : >>> ", e)


# ============================================================
# 3. ELSE BLOCK
# ============================================================
# The else block executes ONLY when the try block
# completes successfully WITHOUT any exception.
#
# In simple words:
#
# try    -> "Try to perform the operation"
# except -> "If something goes wrong, handle the error"
# else   -> "If everything went successfully, do this"
#
# IMPORTANT:
# If an exception occurs in try, Python skips else
# and directly moves to finally.

else:
    print("File Operations done")


# ============================================================
# 4. FINALLY BLOCK
# ============================================================
# The finally block executes ALWAYS.
#
# It runs whether:
#   - try succeeds
#   - try generates an exception
#   - except handles the exception
#
# This makes finally useful for CLEANUP operations,
# such as closing files, database connections, etc.

finally:
    print("I am Closing the file On success or not")

    # Close the file.
    # This releases the file resource.
    file.close()
```
