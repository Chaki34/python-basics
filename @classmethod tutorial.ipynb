```
#classmethod is used to create a method that works with the class instead of an
#object instance. A class method receives the class itself as the first argument using cls.
#It is commonly used to access classvariables, create factory methods and perform operations
#related to the class.

#basically in simle when we need to overload init (construtor).
#Instead, @classmethod is commonly used to provide alternative
#ways of creating an object, and those methods can internally call cls(...), which invokes __init__.
class User :

   def __init__ (self,name,email,phone):
    self.name=name
    self.email=email
    self.phone=phone



  # class method
  # via this we can acive overloading in python
   def print(self):
     print(self.name,self.email,self.phone)

users  = User("debmalya","deb@gmail.com",8923121212)
print(users.name)
print(users.email)

users.print()

```

```
#classmethod is used to create a method that works with the class instead of an
#object instance. A class method receives the class itself as the first argument using cls.
#It is commonly used to access classvariables, create factory methods and perform operations
#related to the class.

#basically in simle when we need to overload init (construtor).
#Instead, @classmethod is commonly used to provide alternative
#ways of creating an object, and those methods can internally call cls(...), which invokes __init__.
class User :

   def __init__ (self,name,email,phone):
    self.name=name
    self.email=email
    self.phone=phone


  #use  @classmethod
  #cls like a pinter

   @classmethod
   def details(cls,name,email,phone):
    s = cls(name,email,phone)
    return s



  # class method

   def print(self):
     print(self.name,self.email,self.phone)


#not create the obj
l = User.details("debmalya","deb@gmail.com",8923121212)
l.print()


```
