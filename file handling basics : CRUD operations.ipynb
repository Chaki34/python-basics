```
# File handling methods

import os

# Length / size of the file in bytes
print("File size:", os.path.getsize("text.txt"), "bytes")

# Check whether file exists
print("File exists:", os.path.exists("text.txt"))

# Check whether it is a file
print("Is it a file:", os.path.isfile("text.txt"))

# Check whether it is a directory
print("Is it a directory:", os.path.isdir("text.txt"))

# Get absolute path of file
print("Absolute path:", os.path.abspath("text.txt"))

# Get file name
print("File name:", os.path.basename("text.txt"))

# Get directory containing the file
print("Directory:", os.path.dirname("text.txt"))

# Get current working directory
print("Current working directory:", os.getcwd())

# List all files and folders in current directory
print("Files and folders:", os.listdir())


# Open file for reading
mfile = open("text.txt", "r")

# Read complete file
print("Complete file:")
print(mfile.read())

# Move cursor back to beginning
mfile.seek(0)

# Read one line
print("First line:")
print(mfile.readline())

# Move cursor back to beginning
mfile.seek(0)

# Read all lines as a list
print("All lines:")
print(mfile.readlines())

# Move file cursor
mfile.seek(0)

# Get current cursor position
print("Current cursor position:", mfile.tell())

# Close file
mfile.close()

# Check whether file is closed
print("File is closed:", mfile.closed)
```
