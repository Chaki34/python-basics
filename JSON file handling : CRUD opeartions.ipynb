```
# JSON file handling

import json


data = {
    "id": "434fds3323",
    "name": "jhon",
    "email": "jhon@gmail.com",
    "wishlist": ["samsung", "iphone", "Vivo"],
    "purchase": {
        "samsung": 2,
        "iphone": 1,
        "Vivo": 1
    }
}

data1 = {
    "id": "434sdsds3323",
    "name": "jhona",
    "email": "jhona@gmail.com",
    "wishlist": ["samsung", "iphone", "Vivo", "oppo"],
    "purchase": {
        "samsung": 2,
        "Vivo": 1
    }
}


# Store multiple objects inside a list
users = [data, data1]


# Write data into JSON file
with open("sample.json", "w") as f:
    json.dump(users, f, indent=4)


# Read JSON data
with open("sample.json", "r") as f:
    result = json.load(f)

print(result)

# separalty stoere a json not inside the list
with open ("sample.json","a") as f :
    json.dump(data,f)
```
