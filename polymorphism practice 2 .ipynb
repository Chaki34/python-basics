```
# task

class DS :

  def roadmap (self):
    print("sys for DS")

class WEBDEV :
  def roadmap (self):
    print("sys for Web Dev")


# method that call two diffrent roadmap method  from diff callses

def fun (list_obj):
  for i in list_obj :
    i.roadmap()

DS_obj = DS()
WEBDEV_obj = WEBDEV()

obj_list  = [DS_obj,WEBDEV_obj]

fun(obj_list)
```

```
# runtime polymorphysam

class  course :

  def roadmap (self):
    print("roadmap from corse")

class DS (course):

  def roadmap (self):
    print("roadmap from DS")

class WEB (course):
  def roadmap(self):
    print("roadmap from  WEB")


a = course()
a.roadmap()
courses = [DS(), WEB()]

for i in courses :
  i.roadmap()  # At runtime, Python checks whether 'i' is a DS or WEBDEV object and calls that class's roadmap() method.
```
