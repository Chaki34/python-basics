```
# Map function


# Map funtion help to apply fun in the items of any itratable data struture

# basically map(fun , itratable) takes as arguments  fun  and another itratable ds

# in treditional way

l = [10,20,30,40]
l2=[]
for i in l :
  l2.append(i**2)

l2


```

```
# but here similar operation done for each element so we alos do this using map funtion

# make a fun that  take a item and make the squere

li = [123,2343,54,12]

make_square = lambda c : c ** 2

print(map(make_square, li))

print(list(map(make_square,li))) # to get itrateble list or what use type wrapping

# more prescisely

print(list(map(lambda c : c **  3 , li)))
```

```
# task

l1 = [1,3,2,4,5]
l2 = [10,34,23,43]

# multiply both same place items with each other and return a new list

def fun(l1, l2):
    new_l = []

    for i in range(min(len(l1), len(l2))):
        new_l.append(l1[i] * l2[i])

    return new_l

print(fun(l1, l2))

```

```
# in simple way

l1 = [1, 3, 2, 4, 5]
l2 = [10, 34, 23, 43]

sum_list = lambda x ,y : x + y

# use map fun

sum  = list(map(sum_list , l1 ,l2))

print(sum)


```

```
# task

string  = "debmalya"

# convert this str into upercase

uppercase_str = lambda s : s.upper()
print(list(map(uppercase_str , string)))

result = "".join(map(uppercase_str, string)) # to get str

print(result)



```
