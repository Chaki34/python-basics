```
# Encapsulation

# task  make a class User and make it encapsulate
# users only set and get data  via methods.

class User :


  # we not make two or more constructor in same class  only one possible


  # constructor
  def __init__(self,id = 0,name = "",age = 0 , role = "N/A") :

   self.__id = id                   # bascally we make the data fields private using ' __ '
   self.__name = name
   self.__age = age
   self.__role = role

# to acccess  data fileds  make getter and setter

  def set_id(self,id):
    self.__id = id

  def get_id(self):
    return self.__id

  def set_name(self,name):
    self.__name = name

  def get_name(self):
    return self.__name

  def set_age(self,age):
    self.__age = age

  def get_age(self):
    return self.__age

  def set_role(self,role):
    self.__role = role

  def get_role(self):
    return self.__role
```

```
# now not get  any filed via  object
ram  = User(1001,"RAM",23,"STAFF - ADMIN")
ram.get_id()
ram.set_id(12)
ram.get_id()

rohim  = User() # by intilize now we make obj wihtout passing args into params

rohim.set_id(1002)
rohim.set_name("RAHIM")
rohim.set_age(21)
rohim.set_role("USER - ADMIN")

print("ID :",rohim.get_id())
print("NAME :",rohim.get_name())
print("AGE ",rohim.get_age())
print("ROLE ",rohim.get_role())




```

```
# task  make a bank transfer using two obj one is sender and another one is  receiver

class Account :

  def __init__ (self,AccountNo = "", name = "" , bal = 0):
    self.__AccountNo = AccountNo
    self.__name = name
    self.__bal = bal

  def set_accNo(self,accNo):
      self.__AccountNo = accNo

  def set_name(self,name):
      self.__name = name

  def get_accNo(self):
      return self.__AccountNo

  def get_name(self):
      return  self.__name

  def get_bal(self):
      return  self.__bal

  def set_bal(self,bal):
      self.__bal = bal;


```

```
# create two objs

sender  = Account("ACC123","Debmalya",1000)

print("=== Snder Details ===")
print()
print("ACC NO : ",sender.get_accNo())
print("NAME :",sender.get_name())
print("BALABCE : ",sender.get_bal())
print()


receiver  = Account("ACC123","Deb",500)

print("=== Receiver Details ===")
print()
print("ACC NO : ",receiver.get_accNo())
print("NAME :",receiver.get_name())
print("BALABCE : ",receiver.get_bal())
print()

# transfer logic


print("AVL BAL :",sender.get_bal())
transfer_am = int(input("Enter your transfer ammount  : $"))

cur_sndr_bl = sender.get_bal() - transfer_am

reciver_total_amount = receiver.get_bal() + transfer_am

sender.set_bal(cur_sndr_bl)
receiver.set_bal(reciver_total_amount)


print("Sender Sent  : ",sender.get_bal())
print("recevier ricived : ",receiver.get_bal())
```
