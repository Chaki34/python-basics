```
from flask import Flask, send_file

from flask import request

from flask import redirect

app = Flask(__name__)

@app.route("/")
def hello():
    return "<h1>Hello World!</h1>"


# <name> is a dynamic URL parameter.
# Whatever value is written after "/" will be passed to the
# 'name' parameter of the hello_name() function.
#
#
# Here, "Debmalya" is passed as the value of 'name'.
@app.route("/<name>")
def hello_name(name):
    return "<h1>Hello {}!</h1>".format(name)


# using query

@app.route("/sum")
def hello_sum():
    # Get query parameters 'a' and 'b' from the URL
    num1 = int(request.args.get("a"))
    num2 = int(request.args.get("b"))

    # Add the two numbers
    result = num1 + num2

    # Display the result as HTML
    return "<h1>SUM IS: {}</h1>".format(result)




@app.route("/details")
def detailsview():
    return "<h1>Details View</h1>"


@app.route("/index")
def appview():
    return send_file("/content/sample_data/index.html")


@app.route("/details2")
def detailview2():
  return redirect("/details")




if __name__ == "__main__":
   app.run(host="0.0.0.0", port=5000)
```
