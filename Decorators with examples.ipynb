```
# decorators

#Decorators are a powerful and elegant feature in Python that let you modify or
#extend the behavior of functions and methods without changing their code.

# A decorator is a design pattern in Python that allows a
#user to add new functionality to an existing object without
#modifying its structure. Decorators are typically applied to
#functions, and they play a crucial role in enhancing or modifying
# the behavior of functions. Traditionally, decorators are placed before
# the definition of a function you want to decorate. In this tutorial, we'll
# demonstrate how to effectively use decorators in Python functions.


# Decorator = adds extra behavior to a function without changing its original code.
# Real-world use: logging, authentication, authorization, caching, validation, timing, etc.
# Example: login_required → checks login before running the function.
# timer → measures function execution time.
# cache → stores results to avoid repeated work.


# make geric function using decorator  that find time complexcity of any function

import time

def time_complexcity(fun):
    def inner_method():
        start_time = time.time()

        fun()

        end_time = time.time()
        print(end_time - start_time)

    return inner_method
```

```

@time_complexcity
def func1():
    print("show func 1 execution time:")

func1()

```

```
@time_complexcity
def func2():
  for i in range (1000):
    pass


func2()
```

```
# if we paramitarized methods

import time

def time_complexcity_for_parm(fun):
    def inner_method(*args, **kwargs):
        start_time = time.time()

        fun(*args, **kwargs)

        end_time = time.time()

        print("Execution time:", end_time - start_time)

    return inner_method




```

```
@time_complexcity_for_parm
def func3(a, b):
    print(int(a + b))


func3(10, 20)
```

```

@time_complexcity_for_parm
def func5(l):
  result  = 0
  for i in l :
   result  = result + i

  print(result)


lis = [10,20,34,23]

func5(lis)
```
