```
from urllib.parse import urljoin

BASE_URL = "https://www.flipkart.com"

products = []
seen = set()

for link in soup.find_all("a", href=True):

    href = link["href"]
    name = link.get_text(" ", strip=True)

    if "/p/" in href and name:

        product_url = urljoin(BASE_URL, href)

        if product_url not in seen:
            seen.add(product_url)

            products.append({
                "name": name,
                "url": product_url
            })

print("Unique products found:", len(products))

for i, product in enumerate(products[:10], 1):
    print(f"\n{i}. {product['name']}")
    print(product["url"])
```

```
product_data = []

for product in products:

    url = product["url"]
    name = product["name"]

    # Find the <a> element for this product
    link = soup.find(
        "a",
        href=lambda x: x and url.split("?")[0].replace(BASE_URL, "") in x
    )

    if not link:
        continue

    # Get the surrounding product section
    parent = link

    for _ in range(4):
        if parent.parent:
            parent = parent.parent

    card_text = parent.get_text(
        " ",
        strip=True
    )

    product_data.append({
        "Product Name": name,
        "Product URL": url,
        "Product Text": card_text
    })

print("Products collected:", len(product_data))

for product in product_data[:5]:
    print("\n--------------------")
    print(product["Product Name"])
    print(product["Product Text"][:1000])
```

```
import re
from urllib.parse import urljoin

results = []

for product in product_data:

    text = product["Product Text"]

    # --------------------------------
    # PRICE
    # --------------------------------
    prices = re.findall(r"₹\s*[\d,]+", text)

    selling_price = prices[0] if len(prices) > 0 else None
    original_price = prices[1] if len(prices) > 1 else None

    # --------------------------------
    # DISCOUNT
    # --------------------------------
    discount_match = re.search(
        r"(\d+)%\s*off",
        text,
        re.IGNORECASE
    )

    discount = (
        discount_match.group(1) + "%"
        if discount_match
        else None
    )

    # --------------------------------
    # RATING
    # --------------------------------
    rating_match = re.search(
        r"\b([0-5](?:\.[0-9])?)\b",
        text
    )

    rating = (
        rating_match.group(1)
        if rating_match
        else None
    )

    # --------------------------------
    # FIND PRODUCT LINK
    # --------------------------------
    product_path = product["Product URL"].replace(
        BASE_URL,
        ""
    )

    link = soup.find(
        "a",
        href=lambda x: (
            x and product_path in x
        )
    )

    # --------------------------------
    # IMAGE URL
    # --------------------------------
    image_url = None

    if link:

        parent = link

        for _ in range(6):
            if parent.parent:
                parent = parent.parent

        images = parent.find_all("img")

        for img in images:

            candidates = [
                img.get("src"),
                img.get("data-src"),
                img.get("data-lazy-src"),
                img.get("data-original"),
                img.get("data-image"),
                img.get("srcset"),
                img.get("data-srcset")
            ]

            for candidate in candidates:

                if not candidate:
                    continue

                if candidate.startswith("data:image"):
                    continue

                candidate_urls = candidate.split(",")

                for candidate_url in candidate_urls:

                    candidate_url = candidate_url.strip()
                    candidate_url = candidate_url.split(" ")[0]

                    if (
                        candidate_url.startswith("https://")
                        and
                        "rukminim2.flixcart.com/image/" in candidate_url
                        and
                        "..." not in candidate_url
                    ):

                        image_url = candidate_url
                        break

                if image_url:
                    break

            if image_url:
                break

    # --------------------------------
    # REVIEW TEXT
    # --------------------------------

    reviews = []

    # Look for common Flipkart review patterns
    review_patterns = [
        r"\b([A-Za-z][A-Za-z0-9 ,.!?'\"()-]{15,300})\s+(?:Helpful|Report Abuse)",
        r"(?:Verified Buyer)\s+([A-Za-z][A-Za-z0-9 ,.!?'\"()-]{15,300})",
    ]

    for pattern in review_patterns:

        matches = re.findall(
            pattern,
            text,
            re.IGNORECASE
        )

        for review in matches:

            review = re.sub(
                r"\s+",
                " ",
                review
            ).strip()

            if (
                len(review) >= 15
                and review not in reviews
            ):
                reviews.append(review)

    # --------------------------------
    # ALTERNATIVE:
    # FIND REVIEW SECTION
    # --------------------------------

    if not reviews:

        review_keywords = [
            "Customer Reviews",
            "Customer Reviews & Ratings",
            "Reviews",
            "Ratings & Reviews"
        ]

        for keyword in review_keywords:

            position = text.lower().find(
                keyword.lower()
            )

            if position != -1:

                review_section = text[
                    position:position + 5000
                ]

                # Find text following rating
                possible_reviews = re.findall(
                    r"\b(?:5|4|3|2|1)\s*(?:\.\d)?\s+"
                    r"([A-Za-z][A-Za-z0-9 ,.!?'\"()-]{20,300})",
                    review_section
                )

                for review in possible_reviews:

                    review = re.sub(
                        r"\s+",
                        " ",
                        review
                    ).strip()

                    if (
                        len(review) >= 20
                        and review not in reviews
                    ):
                        reviews.append(review)

                if reviews:
                    break

    # Keep first 5 actual review texts
    reviews = reviews[:5]

    # --------------------------------
    # CLEAN PRODUCT NAME
    # --------------------------------
    product_name = product["Product Name"]

    product_name = re.sub(
        r"^Add to Compare\s*",
        "",
        product_name,
        flags=re.IGNORECASE
    ).strip()

    product_name = re.sub(
        r"^Currently unavailable\s*",
        "",
        product_name,
        flags=re.IGNORECASE
    ).strip()

    # --------------------------------
    # SAVE
    # --------------------------------
    results.append({
        "Product Name": product_name,
        "Selling Price": selling_price,
        "Original Price": original_price,
        "Discount": discount,
        "Rating": rating,
        "Reviews": " | ".join(reviews) if reviews else None,
        "Product URL": product["Product URL"],
        "Image URL": image_url
    })

print("Extraction completed:", len(results))
```

```
import re
import json
from bs4 import BeautifulSoup

# ============================================================
# USE THE ALREADY LOADED PRODUCT PAGE
# ============================================================

html = response.text

print("=" * 80)
print("EXTRACTING ACTUAL REVIEWS")
print("=" * 80)

print("HTML size:", len(html))


# ============================================================
# HELPER
# ============================================================

def clean_text(value):

    if value is None:
        return None

    if not isinstance(value, str):
        return str(value)

    # Remove HTML tags
    value = re.sub(r"<[^>]+>", " ", value)

    # Remove excessive whitespace
    value = re.sub(r"\s+", " ", value)

    return value.strip()


# ============================================================
# FIND POSSIBLE REVIEW OBJECTS INSIDE FLIPKART JSON
# ============================================================

found_reviews = []


def scan_object(obj):

    # ========================================================
    # DICTIONARY
    # ========================================================

    if isinstance(obj, dict):

        keys_lower = {
            str(k).lower(): k
            for k in obj.keys()
        }


        # ====================================================
        # POSSIBLE REVIEW RATING
        # ====================================================

        rating = None

        for key in [
            "rating",
            "reviewrating",
            "ratingvalue",
            "stars",
            "star"
        ]:

            if key not in keys_lower:
                continue

            value = obj[keys_lower[key]]

            # ------------------------------------------------
            # Numeric rating
            # ------------------------------------------------

            if isinstance(value, (int, float)):

                try:

                    numeric_rating = float(value)

                    if 1 <= numeric_rating <= 5:

                        rating = (
                            int(numeric_rating)
                            if numeric_rating.is_integer()
                            else numeric_rating
                        )

                        break

                except Exception:
                    pass


            # ------------------------------------------------
            # String rating
            # ------------------------------------------------

            elif isinstance(value, str):

                value = clean_text(value)

                if not value:
                    continue

                # Accept only strings that represent
                # an actual rating.
                #
                # Examples:
                #
                # "5"
                # "4"
                # "4.5"
                # "5 stars"
                # "4 star"
                # "Rating: 5"

                rating_patterns = [
                    r"^\s*([1-5](?:\.[0-9])?)\s*$",

                    r"^\s*([1-5](?:\.[0-9])?)\s*stars?\s*$",

                    r"^\s*rating\s*[:\-]?\s*([1-5](?:\.[0-9])?)\s*$"
                ]

                for pattern in rating_patterns:

                    match = re.match(
                        pattern,
                        value,
                        re.IGNORECASE
                    )

                    if match:

                        numeric_rating = float(
                            match.group(1)
                        )

                        rating = (
                            int(numeric_rating)
                            if numeric_rating.is_integer()
                            else numeric_rating
                        )

                        break

                if rating is not None:
                    break


        # ====================================================
        # POSSIBLE ACTUAL REVIEW / COMMENT
        # ====================================================

        review_text = None

        for key in [
            "reviewtext",
            "review_text",
            "review",
            "comment",
            "comments",
            "content",
            "reviewbody",
            "reviewcomment"
        ]:

            if key not in keys_lower:
                continue

            value = obj[keys_lower[key]]

            if isinstance(value, str):

                value = clean_text(value)

                if value and len(value) >= 3:

                    review_text = value

                    break


        # ====================================================
        # POSSIBLE REVIEW TITLE
        # ====================================================

        review_title = None

        for key in [
            "reviewtitle",
            "review_title",
            "title",
            "headline",
            "summary"
        ]:

            if key not in keys_lower:
                continue

            value = obj[keys_lower[key]]

            if isinstance(value, str):

                value = clean_text(value)

                if value and len(value) >= 2:

                    review_title = value

                    break


        # ====================================================
        # POSSIBLE REVIEWER NAME
        # ====================================================

        reviewer_name = None

        for key in [
            "username",
            "user_name",
            "reviewername",
            "reviewer_name",
            "authorname",
            "author_name",
            "displayname",
            "display_name"
        ]:

            if key not in keys_lower:
                continue

            value = obj[keys_lower[key]]

            # -----------------------------------------------
            # Name stored directly as string
            # -----------------------------------------------

            if isinstance(value, str):

                value = clean_text(value)

                if value and len(value) >= 2:

                    reviewer_name = value

                    break


            # -----------------------------------------------
            # Name stored inside nested object
            # -----------------------------------------------

            elif isinstance(value, dict):

                nested_keys = {
                    str(k).lower(): k
                    for k in value.keys()
                }

                for subkey in [
                    "name",
                    "username",
                    "displayname",
                    "display_name"
                ]:

                    if subkey in nested_keys:

                        subvalue = value[
                            nested_keys[subkey]
                        ]

                        if isinstance(subvalue, str):

                            subvalue = clean_text(
                                subvalue
                            )

                            if (
                                subvalue
                                and len(subvalue) >= 2
                            ):

                                reviewer_name = subvalue

                                break

                if reviewer_name:
                    break


        # ====================================================
        # FALLBACK FOR "NAME"
        # ====================================================
        #
        # We check "name" separately because many product
        # objects also contain a "name" field.
        #
        # It is accepted only when the object already looks
        # like a review object.
        # ====================================================

        if (
            reviewer_name is None
            and review_text is not None
            and "name" in keys_lower
        ):

            value = obj[keys_lower["name"]]

            if isinstance(value, str):

                value = clean_text(value)

                if value and len(value) >= 2:

                    reviewer_name = value


        # ====================================================
        # SAVE ONLY REVIEW-LIKE OBJECTS
        # ====================================================

        if review_text:

            # -----------------------------------------------
            # Values that are usually product attributes
            # rather than actual customer reviews
            # -----------------------------------------------

            bad_values = {
                "colour",
                "color",
                "softness",
                "material quality",
                "size",
                "value for money",
                "design",
                "quality",
                "good product",
                "excellent",
                "nice product",
                "worth",
                "worth it"
            }

            review_lower = review_text.lower().strip()

            # -----------------------------------------------
            # Reject obvious product/category data
            # -----------------------------------------------

            if review_lower not in bad_values:

                # -------------------------------------------
                # Avoid extremely long product descriptions
                # -------------------------------------------

                if len(review_text) <= 1000:

                    found_reviews.append({

                        "Rating": rating,

                        "Reviewer Name": reviewer_name,

                        "Review Title": review_title,

                        "Review": review_text

                    })


        # ====================================================
        # CONTINUE RECURSIVELY
        # ====================================================

        for value in obj.values():

            scan_object(value)


    # ========================================================
    # LIST
    # ========================================================

    elif isinstance(obj, list):

        for item in obj:

            scan_object(item)


# ============================================================
# SEARCH ALL SCRIPT TAGS FOR JSON
# ============================================================

soup_product = BeautifulSoup(
    html,
    "html.parser"
)

scripts = soup_product.find_all("script")

print("Script tags found:", len(scripts))


for script in scripts:

    script_text = (
        script.string
        or script.get_text()
    )

    if not script_text:
        continue


    # ========================================================
    # ONLY INSPECT SCRIPTS THAT LOOK LIKE REVIEW DATA
    # ========================================================

    script_lower = script_text.lower()

    if not any(
        word in script_lower
        for word in [
            "review",
            "rating",
            "verified buyer",
            "reviewer"
        ]
    ):

        continue


    # ========================================================
    # TRY COMPLETE JSON
    # ========================================================

    try:

        data = json.loads(script_text)

        scan_object(data)

    except Exception:

        pass


# ============================================================
# REMOVE DUPLICATES
# ============================================================

unique_reviews = []

seen = set()


for item in found_reviews:

    key = (
        str(item.get("Rating")),
        str(item.get("Reviewer Name")),
        str(item.get("Review Title")),
        str(item.get("Review"))
    )

    if key not in seen:

        seen.add(key)

        unique_reviews.append(item)


# ============================================================
# SHOW RESULTS
# ============================================================

print()

print("=" * 80)

print(
    "ACTUAL REVIEWS FOUND:",
    len(unique_reviews)
)

print("=" * 80)


for i, review in enumerate(
    unique_reviews[:20],
    1
):

    print()

    print(f"REVIEW #{i}")

    print("-" * 60)

    print(
        "Rating       :",
        review["Rating"]
    )

    print(
        "Reviewer     :",
        review["Reviewer Name"]
    )

    print(
        "Review Title :",
        review["Review Title"]
    )

    print(
        "Review       :",
        review["Review"]
    )
```

```
import requests
import re
import json
import time
from bs4 import BeautifulSoup


# ============================================================
# HEADERS
# ============================================================

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/151.0.0.0 Safari/537.36"
    ),
    "Accept": (
        "text/html,application/xhtml+xml,"
        "application/xml;q=0.9,image/avif,image/webp,"
        "*/*;q=0.8"
    ),
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": "https://www.flipkart.com/"
}


# ============================================================
# TEXT CLEANER
# ============================================================

def clean_text(value):

    if value is None:
        return None

    if not isinstance(value, str):
        value = str(value)

    value = re.sub(r"<[^>]+>", " ", value)
    value = re.sub(r"\s+", " ", value)

    return value.strip()


# ============================================================
# EXTRACT RATING FROM VALUE
# ============================================================

def extract_rating(value):

    if value is None:
        return None

    # --------------------------------------------------------
    # Numeric
    # --------------------------------------------------------

    if isinstance(value, (int, float)):

        try:

            value = float(value)

            if 1 <= value <= 5:

                if value.is_integer():
                    return int(value)

                return value

        except Exception:
            pass


    # --------------------------------------------------------
    # String
    # --------------------------------------------------------

    if isinstance(value, str):

        value = clean_text(value)

        if not value:
            return None

        patterns = [

            # "5"
            r"^\s*([1-5](?:\.[0-9])?)\s*$",

            # "5 stars"
            r"^\s*([1-5](?:\.[0-9])?)\s*stars?\s*$",

            # "Rating: 5"
            r"rating\s*[:\-]?\s*([1-5](?:\.[0-9])?)",

            # "5/5"
            r"^\s*([1-5](?:\.[0-9])?)\s*/\s*5\s*$"
        ]

        for pattern in patterns:

            match = re.search(
                pattern,
                value,
                re.IGNORECASE
            )

            if match:

                try:

                    number = float(match.group(1))

                    if 1 <= number <= 5:

                        if number.is_integer():
                            return int(number)

                        return number

                except Exception:
                    pass

    return None


# ============================================================
# FIND RATING INSIDE AN OBJECT
# ============================================================

def find_rating(obj):

    if not isinstance(obj, dict):
        return None

    keys_lower = {
        str(k).lower(): k
        for k in obj.keys()
    }


    # --------------------------------------------------------
    # Direct rating fields
    # --------------------------------------------------------

    rating_keys = [

        "rating",
        "reviewrating",
        "review_rating",

        "ratingvalue",
        "rating_value",

        "ratingvalue",

        "stars",
        "star",

        "starrating",
        "star_rating",

        "ratingstar",
        "rating_star",

        "value"
    ]


    for key in rating_keys:

        if key in keys_lower:

            rating = extract_rating(
                obj[keys_lower[key]]
            )

            if rating is not None:
                return rating


    # --------------------------------------------------------
    # Search nested rating object
    # --------------------------------------------------------

    for key, original_key in keys_lower.items():

        value = obj[original_key]

        if isinstance(value, dict):

            nested_keys = {
                str(k).lower(): k
                for k in value.keys()
            }

            for rating_key in [
                "rating",
                "ratingvalue",
                "rating_value",
                "stars",
                "star",
                "value"
            ]:

                if rating_key in nested_keys:

                    rating = extract_rating(
                        value[nested_keys[rating_key]]
                    )

                    if rating is not None:
                        return rating

    return None


# ============================================================
# FIND REVIEW TEXT
# ============================================================

def find_review_text(obj):

    if not isinstance(obj, dict):
        return None

    keys_lower = {
        str(k).lower(): k
        for k in obj.keys()
    }


    review_keys = [

        "reviewtext",
        "review_text",

        "reviewcomment",
        "review_comment",

        "reviewbody",
        "review_body",

        "comment",
        "comments",

        "content"
    ]


    for key in review_keys:

        if key not in keys_lower:
            continue

        value = obj[keys_lower[key]]

        if isinstance(value, str):

            value = clean_text(value)

            if value and len(value) >= 3:

                return value


    return None


# ============================================================
# FIND REVIEW TITLE
# ============================================================

def find_review_title(obj):

    if not isinstance(obj, dict):
        return None

    keys_lower = {
        str(k).lower(): k
        for k in obj.keys()
    }


    title_keys = [

        "reviewtitle",
        "review_title",

        "headline",

        "summary"
    ]


    for key in title_keys:

        if key not in keys_lower:
            continue

        value = obj[keys_lower[key]]

        if isinstance(value, str):

            value = clean_text(value)

            if value and len(value) >= 2:

                return value


    return None


# ============================================================
# FIND REVIEWER NAME
# ============================================================

def find_reviewer_name(obj):

    if not isinstance(obj, dict):
        return None

    keys_lower = {
        str(k).lower(): k
        for k in obj.keys()
    }


    name_keys = [

        "username",
        "user_name",

        "reviewername",
        "reviewer_name",

        "authorname",
        "author_name",

        "displayname",
        "display_name"
    ]


    for key in name_keys:

        if key not in keys_lower:
            continue

        value = obj[keys_lower[key]]


        # ----------------------------------------------------
        # Direct string
        # ----------------------------------------------------

        if isinstance(value, str):

            value = clean_text(value)

            if value and len(value) >= 2:

                return value


        # ----------------------------------------------------
        # Nested object
        # ----------------------------------------------------

        if isinstance(value, dict):

            nested_keys = {
                str(k).lower(): k
                for k in value.keys()
            }

            for subkey in [
                "name",
                "username",
                "displayname",
                "display_name"
            ]:

                if subkey in nested_keys:

                    subvalue = value[
                        nested_keys[subkey]
                    ]

                    if isinstance(subvalue, str):

                        subvalue = clean_text(
                            subvalue
                        )

                        if (
                            subvalue
                            and len(subvalue) >= 2
                        ):

                            return subvalue


    return None


# ============================================================
# CHECK WHETHER OBJECT LOOKS LIKE A REVIEW
# ============================================================

def is_review_object(obj):

    if not isinstance(obj, dict):
        return False

    keys = {
        str(k).lower()
        for k in obj.keys()
    }

    review_indicators = {

        "review",
        "reviewtext",
        "review_text",

        "reviewcomment",
        "review_comment",

        "reviewbody",
        "review_body",

        "comment",
        "content",

        "reviewername",
        "reviewer_name",

        "username",
        "authorname",

        "rating",
        "reviewrating",
        "ratingvalue"
    }

    return bool(keys.intersection(review_indicators))


# ============================================================
# SCAN JSON
# ============================================================

def scan_object(obj, found_reviews):

    # ========================================================
    # DICTIONARY
    # ========================================================

    if isinstance(obj, dict):

        if is_review_object(obj):

            review_text = find_review_text(obj)

            reviewer_name = find_reviewer_name(obj)

            review_title = find_review_title(obj)

            rating = find_rating(obj)


            # ------------------------------------------------
            # IMPORTANT:
            # Search nearby nested objects for rating
            # ------------------------------------------------

            if review_text and rating is None:

                for value in obj.values():

                    if isinstance(value, dict):

                        nested_rating = find_rating(value)

                        if nested_rating is not None:

                            rating = nested_rating

                            break


            # ------------------------------------------------
            # Save actual review
            # ------------------------------------------------

            if review_text:

                bad_values = {

                    "colour",
                    "color",
                    "softness",
                    "material quality",
                    "size",
                    "value for money",
                    "design",
                    "quality",

                    "good product",
                    "excellent",
                    "nice product",
                    "worth",
                    "worth it"
                }


                review_lower = review_text.lower().strip()


                if (
                    review_lower not in bad_values
                    and len(review_text) <= 1000
                ):

                    found_reviews.append({

                        "Rating": rating,

                        "Reviewer Name":
                            reviewer_name,

                        "Review Title":
                            review_title,

                        "Review":
                            review_text
                    })


        # ----------------------------------------------------
        # Continue recursively
        # ----------------------------------------------------

        for value in obj.values():

            scan_object(
                value,
                found_reviews
            )


    # ========================================================
    # LIST
    # ========================================================

    elif isinstance(obj, list):

        for item in obj:

            scan_object(
                item,
                found_reviews
            )


# ============================================================
# EXTRACT REVIEWS FROM ONE PRODUCT PAGE
# ============================================================

def extract_reviews_from_html(html):

    found_reviews = []

    soup_product = BeautifulSoup(
        html,
        "html.parser"
    )

    scripts = soup_product.find_all("script")


    # --------------------------------------------------------
    # Scan JSON scripts
    # --------------------------------------------------------

    for script in scripts:

        script_text = (
            script.string
            or script.get_text()
        )

        if not script_text:
            continue


        script_lower = script_text.lower()


        # Only inspect relevant scripts
        if not any(
            word in script_lower
            for word in [
                "review",
                "rating",
                "verified buyer",
                "reviewer"
            ]
        ):
            continue


        # ----------------------------------------------------
        # Complete JSON
        # ----------------------------------------------------

        try:

            data = json.loads(
                script_text
            )

            scan_object(
                data,
                found_reviews
            )

        except Exception:

            pass


    # --------------------------------------------------------
    # Remove duplicates
    # --------------------------------------------------------

    unique_reviews = []

    seen = set()


    for item in found_reviews:

        key = (

            str(item.get("Rating")),

            str(item.get("Reviewer Name")),

            str(item.get("Review Title")),

            str(item.get("Review"))
        )


        if key not in seen:

            seen.add(key)

            unique_reviews.append(item)


    return unique_reviews


# ============================================================
# AUTOMATIC EXTRACTION FOR ALL PRODUCTS
# ============================================================

all_product_reviews = []

successful_products = 0
products_with_reviews = 0


print()
print("=" * 80)
print("AUTOMATIC PRODUCT REVIEW EXTRACTION")
print("=" * 80)

print(
    "Total products:",
    len(product_data)
)


for index, product in enumerate(
    product_data,
    1
):

    product_name = product.get(
        "Product Name",
        "Unknown Product"
    )

    product_url = product.get(
        "Product URL"
    )


    print()
    print("=" * 80)

    print(
        f"[{index}/{len(product_data)}] "
        f"{product_name[:70]}"
    )

    print("=" * 80)


    if not product_url:

        print("❌ No product URL → Skipping")

        continue


    # ========================================================
    # REQUEST WITH RETRIES
    # ========================================================

    response = None


    for attempt in range(1, 4):

        try:

            print(
                f"Opening product page... "
                f"(attempt {attempt}/3)"
            )


            response = requests.get(

                product_url,

                headers=headers,

                timeout=(8, 20)
            )


            print(
                "Status:",
                response.status_code
            )


            if response.status_code == 200:

                break


            if response.status_code in [
                429,
                500,
                502,
                503,
                504,
                529
            ]:

                print(
                    "⚠️ Temporary server response"
                )

                time.sleep(
                    2 * attempt
                )

                continue


            print(
                "❌ Request failed"
            )

            response = None

            break


        except requests.exceptions.Timeout:

            print(
                "⏱️ Request timed out"
            )

            if attempt < 3:

                print(
                    "Retrying..."
                )

                time.sleep(
                    2 * attempt
                )


        except requests.exceptions.RequestException as e:

            print(
                "❌ Request error:",
                e
            )

            if attempt < 3:

                time.sleep(
                    2 * attempt
                )


    # ========================================================
    # IF PAGE COULD NOT BE LOADED
    # ========================================================

    if response is None or response.status_code != 200:

        print(
            "→ Skipping this product"
        )

        all_product_reviews.append({

            "Product Number": index,

            "Product Name": product_name,

            "Product URL": product_url,

            "Reviews": []

        })

        continue


    successful_products += 1


    print(
        "✅ Product page loaded"
    )

    print(
        "HTML size:",
        len(response.text)
    )


    # ========================================================
    # EXTRACT REVIEWS
    # ========================================================

    print(
        "Extracting reviews..."
    )


    reviews = extract_reviews_from_html(
        response.text
    )


    if reviews:

        products_with_reviews += 1


        print(
            f"\n✅ ACTUAL REVIEWS FOUND: "
            f"{len(reviews)}"
        )


        # ----------------------------------------------------
        # SHOW REVIEWS
        # ----------------------------------------------------

        for review_no, review in enumerate(
            reviews,
            1
        ):

            print()

            print(
                f"REVIEW #{review_no}"
            )

            print(
                "-" * 60
            )

            print(
                "Rating       :",
                review["Rating"]
            )

            print(
                "Reviewer     :",
                review["Reviewer Name"]
            )

            print(
                "Review Title :",
                review["Review Title"]
            )

            print(
                "Review       :",
                review["Review"]
            )


    else:

        print(
            "⚠️ No review data found"
        )


    # ========================================================
    # SAVE PRODUCT + REVIEWS
    # ========================================================

    all_product_reviews.append({

        "Product Number": index,

        "Product Name": product_name,

        "Product URL": product_url,

        "Reviews": reviews

    })


    # ========================================================
    # SMALL DELAY
    # ========================================================

    time.sleep(1)


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 80)
print("AUTOMATION FINISHED")
print("=" * 80)

print(
    "Products successfully loaded:",
    successful_products
)

print(
    "Products containing reviews:",
    products_with_reviews
)

print(
    "Total products processed:",
    len(all_product_reviews)
)

print("=" * 80)
```
