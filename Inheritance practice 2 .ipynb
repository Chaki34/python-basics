```
# simple inheritance

class Parent ():

  def fun (self,):
   print("hello i am parent")

class Child(Parent):
   def fun_c(self,):
        self.fun() # slef couse now fun() is also method of child  by inharit it
        print("I am from Child")
```

```
obj1 = Child()
obj1.fun_c()
```

```
# multi level

class Grandpa :
  name  = "X"

class Father(Grandpa):
  f_name  = "Y"


class Me (Father):
  name6  = "Deb"
  def dis_fun(self):
    print("MY NAME :",self.name6)
    print("FATHER NAME :",self.f_name)
    print("Grandpa name :",self.name)




```

```
obj3  = Me()
obj3.dis_fun()
```

```
# Python directly supports multiple inheritance.

class Father :
  hair  = " long hair"
  eye = " like elephant "
  b_list = ["GOOD","SMILEY","CLAM"]

  def behaviour(self):
    return self.b_list

class Mother :
  skin = "Bright"
  hight = "medium"
  b_lsit = ["GOOD","BEAUTIFUL","CLAM","SOFT HEARTED"]

  def behaviour(self):
    return self.b_lsit
```

```
class Me(Father,Mother):

   def my_info(self):
    print(self.hair)
    print(self.eye)
    print(self.hight)

    # In Python, if both parent classes have a method with the same name, Python uses Method Resolution Order (MRO).

    print(Father.behaviour(self) + Mother.behaviour(self))

```

```
obj  = Me()
obj.my_info()
```
