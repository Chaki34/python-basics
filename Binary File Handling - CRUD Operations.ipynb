```
# Binary File Handling - CRUD Operations

# Color codes
RED = b"\033[91m"
GREEN = b"\033[92m"
YELLOW = b"\033[93m"
BLUE = b"\033[94m"
RESET = b"\033[0m"


# CREATE
with open("testing.bin", "wb") as f:
    f.write(
        GREEN +
        b"Student: Alice, ID: 101, Marks: 85" +
        RESET
    )


# READ
with open("testing.bin", "rb") as f:
    data = f.read()
    print("Read Data:", data)


# UPDATE
with open("testing.bin", "wb") as f:
    f.write(
        YELLOW +
        b"Student: Alice, ID: 101, Marks: 95" +
        RESET
    )


# READ after UPDATE
with open("testing.bin", "rb") as f:
    data = f.read()
    print("Updated Data:", data)




# DELETE
import os

#os.remove("testing.bin")

# print("File deleted successfully")
```
