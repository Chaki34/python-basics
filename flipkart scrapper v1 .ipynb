```
import requests
from bs4 import BeautifulSoup
from urllib.parse import quote

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/151.0.0.0 Safari/537.36"
    ),
    "Accept": (
        "text/html,application/xhtml+xml,"
        "application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8"
    ),
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": "https://www.flipkart.com/"
}
```

```
search_item = input("Enter product: ")

flipkart_url = (
    "https://www.flipkart.com/search?q="
    + quote(search_item)
)

print("URL:", flipkart_url)

response = requests.get(
    flipkart_url,
    headers=headers,
    timeout=30
)

print("Status:", response.status_code)
```

```
with open("flipkart_search.html", "w", encoding="utf-8") as f:
    f.write(response.text)

print("✅ Search page saved")
```

```
soup = BeautifulSoup(response.text, "html.parser")

print(soup.title)
```

```
from urllib.parse import urljoin

BASE_URL = "https://www.flipkart.com"

products = []
seen = set()

for link in soup.find_all("a", href=True):

    href = link["href"]
    name = link.get_text(" ", strip=True)

    if "/p/" in href and name:

        product_url = urljoin(BASE_URL, href)

        if product_url not in seen:
            seen.add(product_url)

            products.append({
                "name": name,
                "url": product_url
            })

print("Unique products found:", len(products))

for i, product in enumerate(products[:10], 1):
    print(f"\n{i}. {product['name']}")
    print(product["url"])
```

```
search_text = soup.get_text(" ", strip=True)

print(search_text[:10000])
```

```
product_data = []

for product in products:

    url = product["url"]
    name = product["name"]

    # Find the <a> element for this product
    link = soup.find(
        "a",
        href=lambda x: x and url.split("?")[0].replace(BASE_URL, "") in x
    )

    if not link:
        continue

    # Get the surrounding product section
    parent = link

    for _ in range(4):
        if parent.parent:
            parent = parent.parent

    card_text = parent.get_text(
        " ",
        strip=True
    )

    product_data.append({
        "Product Name": name,
        "Product URL": url,
        "Product Text": card_text
    })

print("Products collected:", len(product_data))

for product in product_data[:5]:
    print("\n--------------------")
    print(product["Product Name"])
    print(product["Product Text"][:1000])
```

```
import re
from urllib.parse import urljoin

results = []

for product in product_data:

    text = product["Product Text"]

    # --------------------------------
    # PRICE
    # --------------------------------
    prices = re.findall(
        r"₹\s*[\d,]+",
        text
    )

    selling_price = prices[0] if len(prices) > 0 else None
    original_price = prices[1] if len(prices) > 1 else None

    # --------------------------------
    # DISCOUNT
    # --------------------------------
    discount_match = re.search(
        r"(\d+)%\s*off",
        text,
        re.IGNORECASE
    )

    discount = (
        discount_match.group(1) + "%"
        if discount_match
        else None
    )

    # --------------------------------
    # RATING
    # --------------------------------
    rating_match = re.search(
        r"\b([0-5](?:\.[0-9])?)\b",
        text
    )

    rating = (
        rating_match.group(1)
        if rating_match
        else None
    )

    # --------------------------------
    # FIND PRODUCT LINK
    # --------------------------------
    product_path = product["Product URL"].replace(
        BASE_URL,
        ""
    )

    link = soup.find(
        "a",
        href=lambda x: (
            x and product_path in x
        )
    )

    # --------------------------------
    # IMAGE URL
    # --------------------------------
    image_url = None

    if link:

        # Move up to product-card container
        parent = link

        for _ in range(6):

            if parent.parent:
                parent = parent.parent

        # Look at ALL images in the product card
        images = parent.find_all("img")

        for img in images:

            # Check every possible image attribute
            candidates = [
                img.get("src"),
                img.get("data-src"),
                img.get("data-lazy-src"),
                img.get("data-original"),
                img.get("data-image"),
                img.get("srcset"),
                img.get("data-srcset")
            ]

            for candidate in candidates:

                if not candidate:
                    continue

                # Ignore SVG/base64 placeholders
                if candidate.startswith("data:image"):
                    continue

                # Handle srcset:
                # URL1 1x, URL2 2x
                candidate_urls = candidate.split(",")

                for candidate_url in candidate_urls:

                    candidate_url = candidate_url.strip()

                    # Remove srcset size information
                    candidate_url = candidate_url.split(" ")[0]

                    if (
                        candidate_url.startswith("https://")
                        and
                        "rukminim2.flixcart.com/image/" in candidate_url
                    ):

                        # Make sure it is not truncated
                        if "..." not in candidate_url:

                            image_url = candidate_url
                            break

                if image_url:
                    break

            if image_url:
                break

    # --------------------------------
    # CLEAN PRODUCT NAME
    # --------------------------------
    product_name = product["Product Name"]

    product_name = re.sub(
        r"^Add to Compare\s*",
        "",
        product_name,
        flags=re.IGNORECASE
    ).strip()

    product_name = re.sub(
        r"^Currently unavailable\s*",
        "",
        product_name,
        flags=re.IGNORECASE
    ).strip()

    # --------------------------------
    # SAVE
    # --------------------------------
    results.append({
        "Product Name": product_name,
        "Selling Price": selling_price,
        "Original Price": original_price,
        "Discount": discount,
        "Rating": rating,
        "Product URL": product["Product URL"],
        "Image URL": image_url
    })

print("Extraction completed:", len(results))
```
