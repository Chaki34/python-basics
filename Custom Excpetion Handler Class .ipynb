```
# Custom Exception Class

import logging

logging.basicConfig(
    filename="Employee_candidates.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)


class AgeValidate(Exception):

    def __init__(self, msg):
        self.msg = msg
        super().__init__(self.msg)  # properly intilize the custom exception


def Agevalidate(age):

    if age < 18:
        logging.error(f"AGE IS TOO SMALL TO JOIN: {age}")
        print(f"AGE IS TOO SMALL TO JOIN: {age}")
        raise AgeValidate("Age is below 18")

    elif age > 50:
        logging.error(f"AGE IS TOO HIGH TO JOIN: {age}")
        print(f"Oops, not for older one: {age}")
        raise AgeValidate("Age is above 50")

    else:
        logging.info(f"AGE GRANTED: {age}")
        print("Allowed age triggered")


class ValidateEmail (Exception):

   def __init__(self, msg):
        self.msg = msg
        super().__init__(self.msg)



def Validateemail (email) :

  if "@" in email and email.endswith(".com"):
    logging.info(f"Valid Email: {email}")
    print("Valid email")
  else:
    logging.error(f"InValid Email: {email}")
    print("Invalid email")

```

```
try:
    age = int(input("Enter your age: "))

    logging.info(f"INPUT TAKING EVENT TRIGGERED: {age}")
    logging.info(f"VALIDATOR EXECUTED: {age}")

    Agevalidate(age)

except AgeValidate as e:
    print("ERROR:", e)

try:

    try:
        email = str(input("Enter your email: "))

    except Exception as e:
        print("ERROR:", e)

    logging.info(f"INPUT TAKING EVENT TRIGGERED: {email}")
    logging.info(f"VALIDATOR EXECUTED: {email}")

    Validateemail(email)

except ValidateEmail as e:
    logging.error(f"Error except block executed: {email}")
    print("ERROR:", e)

```
