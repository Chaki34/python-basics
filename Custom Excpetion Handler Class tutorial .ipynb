```
# Custom Exception Class

import logging

logging.basicConfig(
    filename="Employee_candidates.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)


# =========================
# Age Custom Exception
# =========================

class AgeValidate(Exception):

    def __init__(self, msg):
        self.msg = msg
        super().__init__(self.msg)


def Agevalidate(age):

    if age < 18:
        logging.error(f"AGE IS TOO SMALL TO JOIN: {age}")
        raise AgeValidate(f"Age {age} is below 18")

    elif age > 50:
        logging.error(f"AGE IS TOO HIGH TO JOIN: {age}")
        raise AgeValidate(f"Age {age} is above 50")

    else:
        logging.info(f"AGE GRANTED: {age}")


# =========================
# Email Custom Exception
# =========================

class ValidateEmail(Exception):

    def __init__(self, msg):
        self.msg = msg
        super().__init__(self.msg)


def Validateemail(email):

    if "@" not in email:
        logging.error(f"Invalid Email: {email}")
        raise ValidateEmail("Email must contain @")

    elif not email.endswith(".com"):
        logging.error(f"Invalid Email: {email}")
        raise ValidateEmail("Email must end with .com")

    else:
        logging.info(f"Valid Email: {email}")


# =========================
# Age Validation
# =========================

try:

    age = int(input("Enter your age: "))

    logging.info(f"INPUT TAKING EVENT TRIGGERED: {age}")
    logging.info(f"VALIDATOR EXECUTED: {age}")

    Agevalidate(age)

except AgeValidate as e:

    logging.error(f"Age exception executed: {e}")
    print("ERROR:", e)


# =========================
# Email Validation
# =========================

try:

    email = input("Enter your email: ")

    logging.info(f"INPUT TAKING EVENT TRIGGERED: {email}")
    logging.info(f"VALIDATOR EXECUTED: {email}")

    Validateemail(email)

except ValidateEmail as e:

    logging.error(f"Email exception executed: {e}")
    print("ERROR:", e)
```
