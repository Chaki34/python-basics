```
# ================= BUFFERED WRITE =================

import io


# 1. BufferedWriter + write()
# write() → Writes ONE bytes object at a time
with open("student.txt", "bw") as f:
    file = io.BufferedWriter(f)

    file.write(b"this is my first line\n")    # Write first line
    file.write(b"this is my second line\n")   # Write second line
    file.write(b"this is my third line\n")    # Write third line

    file.flush()  # Force buffered data to be written to the file


# 2. BufferedWriter + writelines()
# writelines() → Writes MULTIPLE bytes objects from a list/iterable
with open("student.txt", "ba") as f:
    file = io.BufferedWriter(f)

    file.writelines([
        b"dolor sit amet.\n",                    # Line 1
        b"consectetuer adipiscing elit.\n",      # Line 2
        b"Aenean commodo ligula eget dolor.\n"   # Line 3
    ])

    file.flush()  # Ensure all buffered data is written


# ================= QUICK DIFFERENCE =================

# write()      → Writes one bytes object
# writelines() → Writes multiple bytes objects
# flush()      → Sends remaining buffered data to the file
```

```
import io

# ================= BUFFERED READ =================

# 1. read(size) → Reads a specific number of bytes
with open("student.txt", "br") as f:
    file = io.BufferedReader(f)

    data = file.read(10)       # Reads first 10 bytes
    print(data)


# 2. readlines() → Reads all lines and returns them as a list
with open("student.txt", "br") as f:
    file = io.BufferedReader(f)

    data1 = file.readlines()   # Reads ALL lines → returns a list
    print(data1)


# 3. readline() → Reads only ONE line at a time
with open("student.txt", "br") as f:
    file = io.BufferedReader(f)

    data = file.readline()     # Reads only the first line
    print(data)
```
