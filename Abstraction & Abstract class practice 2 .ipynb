```
# Abstraction  & Abstract classes

# ==========================================================
# ABSTRACT CLASS IN PYTHON
# ==========================================================
#
# Python supports abstraction using the built-in 'abc' module.
#
# abc = Abstract Base Class
#
# The abc module allows us to:
# 1. Create abstract classes.
# 2. Create abstract methods.
# 3. Force child classes to implement specific methods.
#
# Abstract classes cannot be instantiated (objects cannot be created).
#
# Example:
#
#   class Animal(ABC):
#       abstractmethod
#       def sound(self):
#           pass
#
#   # Animal()  ❌ Not Allowed
#
#   class Dog(Animal):
#       def sound(self):
#           print("Bark")
#
#   dog = Dog()     # ✅ Allowed
#   dog.sound()
#
# ==========================================================

import abc      # Imports the Abstract Base Class (ABC) module.

# OR

from abc import ABC, abstractmethod ,abstractclassmethod

# ABC             -> Base class used to create an abstract class.
# abstractmethod -> Decorator used to declare an abstract method.

class Body(ABC) :

  @abstractmethod
  def head(self):
    pass
  @abstractmethod
  def hair (self):
    pass
  @abstractmethod
  def eye (self):
    pass
```
